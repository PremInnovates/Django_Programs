```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("advertising_sales_extended.csv")
print(df.head())
```

```python
x=df[["Advertising_Budget"]]
y=df["Sales"]
```

```python
print("Advertising_Budget Summary :")
print(x.describe())
print("\n Sales Summary :")
print(y.describe())
```

```python
plt.scatter(x,y)
plt.xlabel("Advertising Budget")
plt.ylabel("Sales")
plt.title("Advertising Budget vs Sales")
plt.show()
```
