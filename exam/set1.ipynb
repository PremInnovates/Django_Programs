```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv("customer.csv")
print(df.head(15))
```

```python
sorted_age=df.sort_values(by="Age")
print(sorted_age)
```

```python
df_clean=df.dropna()
print(df_clean)
```

```python
plt.figure()
plt.plot(df_clean["Age"],df_clean["Balance"])
plt.xlabel("Age")
plt.ylabel("Balance")
plt.title("Age Vs Balance")
plt.show()
```

```python
sort=df.sort_values(by="CreditScore",ascending=False).head(15)
print(sort)
```

```python
geo_counts=df["Geography"].value_counts()
plt.figure()
plt.pie(geo_counts,labels=geo_counts.index,colors=["red","blue","green"],autopct="%1.1f%%")
plt.title("Customer Discribution by Geography")
plt.show()
```

```python
high_salary=df[df["EstimatedSalary"]>75000]
print(high_salary)
```

```python
mean_balance=df["Balance"].mean()
print(mean_balance)
```

```python
from itertools import permutations

credit_list=df["CreditScore"].head(3).tolist()
perm=list(permutations(credit_list))
print(perm)
```

```python
geo_numeric=df["Geography"].astype("category").cat.codes
plt.figure()
plt.scatter(geo_numeric,df["Balance"])
plt.xlabel("Geometry (Encoded)")
plt.ylabel("Balance")
plt.title("Geometry vs Balance")
plt.show()
```
