```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("pokemon_data.csv")
print(df.tail(5))
```

```python
df["Type2"]=df["Type2"].fillna("FILLED")
print(df["Type2"])
```

```python
Defense_mean=df["Defense"].mean
print(Defense_mean)
```

```python
average_values=df[["HP","Attack","Defense"]].mean()
print("Average Values :\n",average_values)
```

```python
hp_sum=df["HP"].sum()
hp_count=len(df["HP"])
hp_avg=hp_sum/hp_count
print("HP Average :",hp_avg)


Attach_sum=df["Attack"].sum()
Attach_count=len(df["Attack"])
Attach_avg=Attach_sum/Attach_count
print("Attach Average :",Attach_avg)

defense_sum=df["Defense"].sum()
defense_count=len(df["Defense"])
defense_avg=defense_sum/defense_count
print("Defense Average :",defense_avg)
```

```python
plt.figure()

plt.bar(df["Name"].head(10),df["Attack"].head(10))
plt.xticks(rotation=90)
plt.xlabel("name")
plt.ylabel("Attack")
plt.title("Pokemon Name vs Attack")
plt.show()
```

```python
legendary_pokemon=df[df["Legendary"]==True]
print(legendary_pokemon)
```

```python
plt.hist(df["Generation"],bins=10)
plt.xlabel("Generation")
plt.ylabel("Frequency")
plt.title("Generation Distribution")
plt.show()
```

```python
std_speed=df["Speed"].std()
print("Standard Deviation of Speed :",std_speed)
```

```python
plt.scatter(df["HP"],df["Speed"])
plt.xlabel("HP")
plt.ylabel("Speed")
plt.title("HP vs Spped")
plt.show()
```
